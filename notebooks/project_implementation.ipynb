This notebook documents the project input and delegates the repeatable training workflow to the scripts under `src/mlops_pipeline`. Place `churn.csv` and, optionally, `new_churn.csv` in `data/raw` before running.

```python
from pathlib import Path
import pandas as pd

data_path = Path('../data/raw/churn.csv')
if not data_path.exists():
    raise FileNotFoundError(f'Required dataset is missing: {data_path.resolve()}')
df = pd.read_csv(data_path)
print(f'Dataset shape: {df.shape}')
df.head()
```

```python
required_columns = {
    'customerID', 'gender', 'SeniorCitizen', 'Partner', 'Dependents',
    'tenure', 'PhoneService', 'MultipleLines', 'InternetService',
    'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport',
    'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling',
    'PaymentMethod', 'MonthlyCharges', 'TotalCharges', 'Churn'
}
missing = required_columns - set(df.columns)
if missing:
    raise ValueError(f'Missing expected columns: {sorted(missing)}')
df['Churn'].value_counts(dropna=False)
```

Run the reproducible pipeline from the repository root with `python src/mlops_pipeline/churn_prediction_pipeline.py`. Generated datasets, models, reports, and plots are written by pipeline execution and are not checked into this notebook.
